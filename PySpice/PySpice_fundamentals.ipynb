```python
from PySpice.Spice.Netlist import Circuit
import PySpice
import numpy as np
import matplotlib.pyplot as plt
```

```python
circuit=Circuit("voltage Divider")
circuit.V("input","vin",circuit.gnd,10)
circuit.R(1,"vin","vout",1000)
circuit.R(2,"vout",Circuit.gnd,1000)

print(circuit)
```

```python
from PySpice.Spice.Netlist import Circuit

circuit = Circuit('Voltage Divider')

circuit.V('input', 'Vin', circuit.gnd, 10)

circuit.R(1, 'Vin', 'Vout', 1000)
circuit.R(2, 'Vout', circuit.gnd, 1000)
simulator=circuit.simulator()
```
