```python
import numpy as np
import matplotlib.pyplot as plt
import sympy as sp
from sympy.abc import A,B
```

```python
Wc=2.4e6*np.pi
Ws=2e3*np.pi
Ac=100
As=50
t=np.linspace(0,1/1e3,1000)
Yc=Ac*np.sin(Wc*t);
Ys=As*np.sin(Ws*t)

#modulated signal
m=As/Ac;
Amplitude=Ac+m*Ac*np.cos(Ws*t)
Ym=Amplitude*np.cos(Wc*t)
```

```python
_,ax=plt.subplots(ncols=1,nrows=3,figsize=(12,8))
ax[0].plot(t,Ys)
ax[0].set_title("Signal wave")
ax[1].plot(t,Yc)
ax[1].set_title("Carrier wave")
ax[2].plot(t,Ym)
ax[2].set_title("Amplitude modulated wave")
for ax in ax:
    ax.grid(True,axis='y',alpha=0.4)
    plt.tight_layout()
    ax.set_ylabel("Amplitude")
    ax.set_xlabel("Time")
```
