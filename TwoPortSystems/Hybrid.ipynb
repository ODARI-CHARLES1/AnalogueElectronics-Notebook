# Hybrid Parameters (h-Parameters)

![h-parameter-network](images/Readme/h-parameter-network-circuit.png)

The hybrid parameter model expresses a two-port network using input current and output voltage as independent variables:

$
V_1 = h_{11}I_1 + h_{12}V_2
$

$
I_2 = h_{21}I_1 + h_{22}V_2
$

This representation is commonly used for transistor small-signal analysis and amplifier design.

```python
import sympy as sp
import warnings

warnings.filterwarnings("ignore")
```

```python
v1,v2,i1,i2=sp.symbols("v1,v2,i1,i2")
```

```python
from sympy import symbols, Eq, solve

V1, V2, I1, I2 = symbols('V1 V2 I1 I2')

eq1 = Eq(V1, 8*I1 + 6*I2)
eq2 = Eq(V2, 6*I1 + 9*I2)

sol = solve((eq1, eq2), (V1, I2))

print("V1 =", sol[V1])
print("I2 =", sol[I2])
```

```python
sp.simplify(sol[I2])
```
