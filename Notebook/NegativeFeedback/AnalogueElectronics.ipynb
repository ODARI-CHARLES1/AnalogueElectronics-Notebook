```python
import numpy as np
import matplotlib.pyplot as plt

t = np.linspace(0,1,1000)

# Input signal
x = np.sin(2*np.pi*5*t)

# Amplifier noise
noise = 0.5*np.random.randn(len(t))

G = 10
H = 0.9

# Without feedback
y_no_feedback = G*x + noise

# With negative feedback
y_feedback = (G/(1+G*H))*x + noise/(1+G*H)

plt.figure(figsize=(10,4))

plt.plot(t,x,label="Input signal")
plt.plot(t,y_no_feedback,label="Amplifier output (with noise)")
plt.plot(t,y_feedback,label="With negative feedback",linewidth=2)

plt.legend()
plt.xlabel("Time")
plt.ylabel("Amplitude")
plt.title("Noise Reduction Using Negative Feedback")
plt.grid()

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Time axis
t = np.linspace(0, 1, 1000)

# Input signal
x = np.sin(2*np.pi*5*t)

# Noise generated in amplifier
noise = 0.5*np.random.randn(len(t))

# Amplifier parameters
G = 10
H = 0.9

# Without feedback
y_no_feedback = G*x + noise

# With negative feedback
y_feedback = (G/(1+G*H))*x - noise/(1+G*H)

fig,ax=plt.subplots(1,3,figsize=(20,5))
ax[0].plot(t,x,label="Input signal")
ax[0].set_title("Input signal")
ax[1].plot(t,y_no_feedback,color='red',label="Amplifier output (with noise)")
ax[1].set_title("Amplifer output (with noise)")
ax[2].plot(t,y_feedback,color='green',label="With negative feedback",linewidth=2)
ax[2].set_title("With negative feedback")

for ax in ax.flat:
    ax.set(xlabel='Time', ylabel='Amplitude')
    ax.grid(True,axis="y")
    ax.legend()
# --------------------------
```

```python
import numpy as np

w = 2*np.pi*5
Gain = 10  # Gain of the amplifier
Mv = 0.1   # Fraction of output signal

t = np.linspace(0,1,1000)
noise = np.random.randn(len(t))*0.05

input_nf = np.sin(w*t) + noise

# output with no feedback
output_nf = Gain * input_nf

# Gain with negative feedback
Gain_f = Gain/(1 + Gain*Mv)

# feedback signal (180° phase shift)
feedback = Mv * np.max(output_nf) * np.sin(w*t - np.pi)

# input signal with feedback
input_f = input_nf-feedback

# output with feedback
output_f = Gain_f * input_f

print(np.max(input_f))

print(f"Gain without feedback: {Gain}")
print(f"Gain with feedback: {Gain_f}")
print(f"Input signal without feedback: {np.round(np.max(input_nf),2)} sin(w*t)")
print(f"Input signal with feedback: {np.round(np.max(input_f),2)} sin(w*t - π)")
```
