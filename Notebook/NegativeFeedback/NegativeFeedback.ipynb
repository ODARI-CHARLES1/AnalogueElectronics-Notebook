```python
import numpy as np
import matplotlib.pyplot as plt

t = np.linspace(0,1,1000)

# Input signal
x = np.sin(2*np.pi*5*t)

# Amplifier noise
noise = 0.5*np.random.randn(len(t))

G = 10
H = 0.9

# Without feedback
y_no_feedback = G*x + noise

# With negative feedback
y_feedback = (G/(1+G*H))*x + noise/(1+G*H)

plt.figure(figsize=(10,6))

plt.plot(t,x,label="Input signal")
plt.plot(t,y_no_feedback,label="Amplifier output (with noise)")
plt.plot(t,y_feedback,label="With negative feedback",linewidth=2)

plt.legend()
plt.xlabel("Time")
plt.ylabel("Amplitude")
plt.title("Noise Reduction Using Negative Feedback")
plt.grid()

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Time axis
t = np.linspace(0, 1, 1000)

# Input signal
x = np.sin(2*np.pi*5*t)

# Noise generated in amplifier
noise = 0.5*np.random.randn(len(t))

# Amplifier parameters
G = 10
H = 0.9

# Without feedback
y_no_feedback = G*x + noise

# With negative feedback
y_feedback = (G/(1+G*H))*x - noise/(1+G*H)

# --------------------------
# Figure 1: Input signal
plt.figure(figsize=(8,4))
plt.plot(t, x, color='green')
plt.title("Input Signal")
plt.xlabel("Time")
plt.ylabel("Amplitude")
plt.grid(True)
plt.show()

# --------------------------
# Figure 2: Amplifier output with noise
plt.figure(figsize=(8,4))
plt.plot(t, y_no_feedback, color='red')
plt.title("Amplifier Output with Noise (No Feedback)")
plt.xlabel("Time")
plt.ylabel("Amplitude")
plt.grid(True)
plt.show()

# --------------------------
# Figure 3: Output with negative feedback
plt.figure(figsize=(8,4))
plt.plot(t, y_feedback, color='blue')
plt.title("Output with Negative Feedback")
plt.xlabel("Time")
plt.ylabel("Amplitude")
plt.grid(True)
plt.show()
```
